This transformation flips the image horizontally using OpenCV. Flipping is often used during data augmentation to improve model generalization and increase dataset diversity.

```python
# Get image dimensions
(h, w) = image_rgb.shape[:2]

# Define center point
center = (w // 2, h // 2)

# Create rotation matrix
rotation_matrix = cv2.getRotationMatrix2D(center, 45, 1.0)

# Apply rotation
rotated_image = cv2.warpAffine(image_rgb, rotation_matrix, (w, h))

# Display rotated image
plt.imshow(rotated_image)
plt.title("Rotated Image")
plt.axis("off")
plt.show()
```
